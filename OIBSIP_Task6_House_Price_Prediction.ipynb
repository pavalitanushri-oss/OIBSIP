```python
# ============================================================
# OASIS INFOBYTE - DATA ANALYTICS TASK 6
# HOUSE PRICE PREDICTION USING LINEAR REGRESSION
# ============================================================

!pip -q install openpyxl

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import fetch_openml
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.metrics import mean_squared_error, r2_score


# ============================================================
# 1. LOAD HOUSE PRICES DATASET
# ============================================================

print("Loading House Prices dataset...")

try:
    data = fetch_openml(
        name="house_prices",
        as_frame=True,
        parser="auto"
    )

    df = data.frame.copy()

except Exception as e:

    print("OpenML loading failed. Trying alternative source...")

    url = "https://raw.githubusercontent.com/selva86/datasets/master/BostonHousing.csv"

    df = pd.read_csv(url)

    # Create a target column compatible with the analysis
    df = df.rename(
        columns={"medv": "SalePrice"}
    )

print("\nDataset loaded successfully!")

print("Shape:", df.shape)

display(df.head())


# ============================================================
# 2. BASIC DATASET INFORMATION
# ============================================================

print("\nDataset Information")
print("===================")

print("\nRows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)


# ============================================================
# 3. TARGET VARIABLE
# ============================================================

target = "SalePrice"

if target not in df.columns:

    possible_targets = [
        col for col in df.columns
        if col.lower() in [
            "saleprice",
            "price",
            "medv"
        ]
    ]

    if len(possible_targets) > 0:
        target = possible_targets[0]

print("\nTarget variable:", target)


# ============================================================
# 4. MISSING VALUE ANALYSIS
# ============================================================

print("\nMissing values:")
missing_values = df.isnull().sum()

missing_values = (
    missing_values[
        missing_values > 0
    ]
    .sort_values(
        ascending=False
    )
)

display(
    missing_values.head(20)
)


# ============================================================
# 5. DESCRIPTIVE STATISTICS
# ============================================================

print("\nDescriptive statistics:")

display(
    df.describe(
        include="all"
    ).T.head(30)
)


# ============================================================
# 6. TARGET DISTRIBUTION
# ============================================================

plt.figure(
    figsize=(10,6)
)

sns.histplot(
    df[target],
    kde=True
)

plt.title(
    "Distribution of House Prices"
)

plt.xlabel(
    "House Price"
)

plt.ylabel(
    "Frequency"
)

plt.tight_layout()

plt.show()


# ============================================================
# 7. FEATURE SELECTION
# ============================================================

X = df.drop(
    columns=[target]
)

y = pd.to_numeric(
    df[target],
    errors="coerce"
)

# Remove rows where target is missing
valid_target = y.notna()

X = X.loc[
    valid_target
].copy()

y = y.loc[
    valid_target
].copy()

print("\nFeature Selection Discussion")
print("============================")

print("""
Important predictors for house prices may include:

1. Overall quality of the property
2. Living area / floor area
3. Number of rooms
4. Number of bedrooms
5. Property age
6. Location / neighbourhood
7. Garage and basement features

These variables can influence house prices because
larger, newer, better-quality and better-located
properties generally have higher market values.
""")


# ============================================================
# 8. IDENTIFY NUMERIC AND CATEGORICAL FEATURES
# ============================================================

numeric_features = X.select_dtypes(
    include=["int64", "float64", "int32", "float32"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("\nNumeric features:", len(numeric_features))
print("Categorical features:", len(categorical_features))


# ============================================================
# 9. TRAIN / TEST SPLIT - 80/20
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])


# ============================================================
# 10. PREPROCESSING
# ============================================================

numeric_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        )
    ]
)

categorical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            numeric_transformer,
            numeric_features
        ),
        (
            "categorical",
            categorical_transformer,
            categorical_features
        )
    ],
    remainder="drop"
)


# ============================================================
# 11. LINEAR REGRESSION MODEL
# ============================================================

linear_model = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "model",
            LinearRegression()
        )
    ]
)

print("\nTraining Linear Regression model...")

linear_model.fit(
    X_train,
    y_train
)

print("Model trained successfully!")


# ============================================================
# 12. PREDICTIONS
# ============================================================

y_pred = linear_model.predict(
    X_test
)

print("\nFirst 10 predictions:")

prediction_table = pd.DataFrame({
    "Actual Price": y_test.values[:10],
    "Predicted Price": y_pred[:10]
})

display(
    prediction_table
)


# ============================================================
# 13. MODEL EVALUATION
# ============================================================

mse = mean_squared_error(
    y_test,
    y_pred
)

rmse = np.sqrt(
    mse
)

r2 = r2_score(
    y_test,
    y_pred
)

print("\n================================================")
print("LINEAR REGRESSION PERFORMANCE")
print("================================================")

print(
    "Mean Squared Error (MSE):",
    round(mse, 2)
)

print(
    "Root Mean Squared Error (RMSE):",
    round(rmse, 2)
)

print(
    "R² Score:",
    round(r2, 4)
)


# ============================================================
# 14. CORRELATION HEATMAP
# ============================================================

numeric_df = df.select_dtypes(
    include=np.number
)

if target in numeric_df.columns:

    correlation = (
        numeric_df
        .corr()[target]
        .sort_values(
            ascending=False
        )
    )

    print("\nFeatures most correlated with price:")

    display(
        correlation.head(15)
    )

    plt.figure(
        figsize=(12,8)
    )

    sns.heatmap(
        numeric_df.corr(),
        cmap="coolwarm",
        center=0
    )

    plt.title(
        "Correlation Heatmap"
    )

    plt.tight_layout()

    plt.show()


# ============================================================
# 15. ACTUAL VS PREDICTED PRICES
# ============================================================

plt.figure(
    figsize=(8,6)
)

plt.scatter(
    y_test,
    y_pred,
    alpha=0.6
)

minimum = min(
    y_test.min(),
    y_pred.min()
)

maximum = max(
    y_test.max(),
    y_pred.max()
)

plt.plot(
    [minimum, maximum],
    [minimum, maximum]
)

plt.title(
    "Actual Prices vs Predicted Prices"
)

plt.xlabel(
    "Actual Price"
)

plt.ylabel(
    "Predicted Price"
)

plt.tight_layout()

plt.show()


# ============================================================
# 16. RESIDUAL ANALYSIS
# ============================================================

residuals = (
    y_test.values
    - y_pred
)

plt.figure(
    figsize=(10,6)
)

plt.scatter(
    y_pred,
    residuals,
    alpha=0.6
)

plt.axhline(
    y=0
)

plt.title(
    "Residual Plot"
)

plt.xlabel(
    "Predicted Price"
)

plt.ylabel(
    "Residual"
)

plt.tight_layout()

plt.show()


# ============================================================
# 17. RESIDUAL DISTRIBUTION
# ============================================================

plt.figure(
    figsize=(10,6)
)

sns.histplot(
    residuals,
    kde=True
)

plt.title(
    "Distribution of Residuals"
)

plt.xlabel(
    "Residual"
)

plt.ylabel(
    "Frequency"
)

plt.tight_layout()

plt.show()


# ============================================================
# 18. COEFFICIENT ANALYSIS
# ============================================================

try:

    fitted_preprocessor = (
        linear_model
        .named_steps["preprocessor"]
    )

    fitted_model = (
        linear_model
        .named_steps["model"]
    )

    feature_names = (
        fitted_preprocessor
        .get_feature_names_out()
    )

    coefficients = (
        fitted_model.coef_
    )

    coefficient_df = pd.DataFrame({

        "Feature":
            feature_names,

        "Coefficient":
            coefficients,

        "Absolute Impact":
            np.abs(coefficients)

    })

    coefficient_df = (
        coefficient_df
        .sort_values(
            "Absolute Impact",
            ascending=False
        )
    )

    print("\nTop features by coefficient impact:")

    display(
        coefficient_df.head(20)
    )

    # Top positive and negative
    print("\nTop Positive Features:")

    display(
        coefficient_df
        .sort_values(
            "Coefficient",
            ascending=False
        )
        .head(10)
    )

    print("\nTop Negative Features:")

    display(
        coefficient_df
        .sort_values(
            "Coefficient",
            ascending=True
        )
        .head(10)
    )

except Exception as e:

    print(
        "Coefficient analysis could not be displayed:",
        e
    )


# ============================================================
# 19. COEFFICIENT VISUALIZATION
# ============================================================

try:

    top_coefficients = (
        coefficient_df
        .head(15)
        .sort_values(
            "Coefficient"
        )
    )

    plt.figure(
        figsize=(10,7)
    )

    plt.barh(
        top_coefficients["Feature"],
        top_coefficients["Coefficient"]
    )

    plt.title(
        "Top 15 Feature Coefficients"
    )

    plt.xlabel(
        "Coefficient"
    )

    plt.ylabel(
        "Feature"
    )

    plt.tight_layout()

    plt.show()

except:
    pass


# ============================================================
# 20. RIDGE REGRESSION - BONUS COMPARISON
# ============================================================

ridge_model = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "model",
            Ridge(alpha=10)
        )
    ]
)

ridge_model.fit(
    X_train,
    y_train
)

ridge_pred = ridge_model.predict(
    X_test
)

ridge_mse = mean_squared_error(
    y_test,
    ridge_pred
)

ridge_rmse = np.sqrt(
    ridge_mse
)

ridge_r2 = r2_score(
    y_test,
    ridge_pred
)


# ============================================================
# 21. MODEL COMPARISON
# ============================================================

comparison = pd.DataFrame({

    "Model": [
        "Linear Regression",
        "Ridge Regression"
    ],

    "MSE": [
        mse,
        ridge_mse
    ],

    "RMSE": [
        rmse,
        ridge_rmse
    ],

    "R2 Score": [
        r2,
        ridge_r2
    ]

})

print("\n================================================")
print("MODEL COMPARISON")
print("================================================")

display(
    comparison
)


# ============================================================
# 22. MODEL COMPARISON GRAPH
# ============================================================

plt.figure(
    figsize=(9,6)
)

plt.bar(
    comparison["Model"],
    comparison["R2 Score"]
)

plt.title(
    "R² Score Comparison"
)

plt.xlabel(
    "Model"
)

plt.ylabel(
    "R² Score"
)

plt.tight_layout()

plt.show()


# ============================================================
# 23. FINAL CONCLUSION
# ============================================================

best_model = comparison.loc[
    comparison["R2 Score"].idxmax(),
    "Model"
]

best_r2 = comparison["R2 Score"].max()

print("\n================================================")
print("TASK 6 FINAL CONCLUSION")
print("================================================")

print(
    "Best performing model:",
    best_model
)

print(
    "Best R² Score:",
    round(best_r2, 4)
)

print("""
Key data-driven insights:

1. House price is strongly influenced by important
   property characteristics such as quality, size,
   rooms and location-related variables.

2. The actual-vs-predicted plot shows how closely
   the model predictions follow real house prices.

3. Residual analysis helps identify systematic errors
   and areas where the linear model may not capture
   complex relationships.

Linear Regression provides an interpretable baseline,
while Ridge Regression can help control large
coefficients and reduce overfitting.
""")


# ============================================================
# 24. SAVE RESULTS
# ============================================================

prediction_output = pd.DataFrame({

    "Actual_Price":
        y_test.values,

    "Predicted_Price":
        y_pred,

    "Residual":
        residuals

})

prediction_output.to_csv(
    "house_price_predictions.csv",
    index=False
)

comparison.to_csv(
    "house_price_model_comparison.csv",
    index=False
)

print("\nFiles saved:")
print("1. house_price_predictions.csv")
print("2. house_price_model_comparison.csv")

print("\n================================================")
print("TASK 6 COMPLETED SUCCESSFULLY!")
print("================================================")
```
